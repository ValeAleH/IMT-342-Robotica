Este notebook implementa las matrices de rotación en los ejes X, Y y Z, calcula la matriz final $^A R_B$ y verifica sus propiedades de ortogonalidad e isometría.

## 1. Importación de librerías

```python
import numpy as np
```

## 2. Funciones de rotación

Se definen funciones modulares para construir matrices de rotación alrededor de los ejes X, Y y Z. Los ángulos de entrada están expresados en grados.

```python
def rot_x(deg):
    """Matriz de rotación alrededor del eje X."""
    rad = np.deg2rad(deg)
    c = np.cos(rad)
    s = np.sin(rad)

    return np.array([
        [1, 0, 0],
        [0, c, -s],
        [0, s, c]
    ])


def rot_y(deg):
    """Matriz de rotación alrededor del eje Y."""
    rad = np.deg2rad(deg)
    c = np.cos(rad)
    s = np.sin(rad)

    return np.array([
        [c, 0, s],
        [0, 1, 0],
        [-s, 0, c]
    ])


def rot_z(deg):
    """Matriz de rotación alrededor del eje Z."""
    rad = np.deg2rad(deg)
    c = np.cos(rad)
    s = np.sin(rad)

    return np.array([
        [c, -s, 0],
        [s, c, 0],
        [0, 0, 1]
    ])
```

## 3. Ejercicio 2: Composición de rotaciones

La secuencia indicada es:

1. $45^\circ$ alrededor de $X_A$ fijo → **pre-multiplicación**.
2. $90^\circ$ alrededor de $Z_B$ móvil → **post-multiplicación**.
3. $-45^\circ$ alrededor de $Y_A$ fijo → **pre-multiplicación**.

Por lo tanto:

$$^A R_B = R_y(-45^\circ) R_x(45^\circ) R_z(90^\circ)$$

```python
# Matrices correspondientes a cada movimiento
Rx_45 = rot_x(45)
Rz_90 = rot_z(90)
Ry_m45 = rot_y(-45)

# Composición final
R_AB = Ry_m45 @ Rx_45 @ Rz_90

print("Matriz final ^A R_B:")
print(np.round(R_AB, 6))
```

## 4. Validación de ortogonalidad

Una matriz de rotación debe cumplir:

$$R^T R = I$$

Se calcula el error utilizando la norma de Frobenius:

$$\|R^T R-I\|_F < 10^{-15}$$

```python
I = np.eye(3)

error_ortogonalidad = np.linalg.norm(
    R_AB.T @ R_AB - I,
    ord='fro'
)

print("R^T R =")
print(np.round(R_AB.T @ R_AB, 10))

print("\nError de ortogonalidad (norma de Frobenius):")
print(error_ortogonalidad)

if error_ortogonalidad < 1e-15:
    print("\nVALIDACIÓN EXITOSA: R^T R ≈ I")
else:
    print("\nLa matriz no cumple la tolerancia requerida.")
```

## 5. Validación de isometría

Se utiliza el vector:

$$p^B = [2,-1,4]^T$$

Se calcula:

$$p^A = {}^A R_B p^B$$

Finalmente se verifica que una matriz de rotación conserva la norma:

$$\|p^A\| - \|p^B\| \approx 0$$

```python
p_B = np.array([2, -1, 4])

# Transformación del vector desde B hacia A
p_A = R_AB @ p_B

print("p^B =")
print(p_B)

print("\np^A = ^A R_B @ p^B =")
print(np.round(p_A, 6))

# Cálculo de las normas
norma_pB = np.linalg.norm(p_B)
norma_pA = np.linalg.norm(p_A)
diferencia_normas = norma_pA - norma_pB

print("\n||p^B|| =", norma_pB)
print("||p^A|| =", norma_pA)
print("||p^A|| - ||p^B|| =", diferencia_normas)

if np.isclose(norma_pA, norma_pB, atol=1e-15):
    print("\nVALIDACIÓN EXITOSA: La rotación conserva la norma del vector.")
else:
    print("\nLa conservación de la norma no cumple la tolerancia establecida.")
```
